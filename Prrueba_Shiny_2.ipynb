```python
# Instalar dependencias necesarias
!pip install plotly pandas ipywidgets

# Importar bibliotecas
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np

# Habilitar widgets en Colab
from google.colab import output
output.enable_custom_widget_manager()

# ============================================================
# 1. Cargar los datos - Esta parte debe modificarse para Colab
# ============================================================
# En Colab, sube el archivo Excel primero o usa este enlace:
# from google.colab import files
# uploaded = files.upload()  # Esto abrirá un selector de archivos

# Luego lee el archivo:
# df_c3 = pd.read_excel("EPA_Serie_histórica_2026_1T.xlsx", sheet_name="C.3")
# df_31 = pd.read_excel("EPA_Serie_histórica_2026_1T.xlsx", sheet_name="3.1")

# Si quieres usar datos de ejemplo sin subir el archivo, usa el código de ejemplo que ya tienes

# ============================================================
# 2. Función para cargar datos desde Excel (en Colab)
# ============================================================
def cargar_datos_desde_excel():
    """
    Función para cargar datos desde Excel en Colab
    """
    from google.colab import files
    import io

    print("Por favor, sube el archivo 'EPA_Serie_histórica_2026_1T.xlsx'")
    uploaded = files.upload()

    for filename in uploaded.keys():
        # Leer las hojas
        df_c3 = pd.read_excel(io.BytesIO(uploaded[filename]), sheet_name="C.3", skiprows=9)
        df_31 = pd.read_excel(io.BytesIO(uploaded[filename]), sheet_name="3.1", skiprows=9)

        # Limpiar los datos (esto depende de la estructura real del Excel)
        # Omitimos las filas vacías
        df_c3 = df_c3.dropna(how='all', subset=df_c3.columns[1:])
        df_31 = df_31.dropna(how='all', subset=df_31.columns[1:])

        return df_c3, df_31

    return None, None

# ============================================================
# 3. Si no tienes el archivo, usa datos de ejemplo
# ============================================================
def crear_datos_ejemplo():
    """Crea datos de ejemplo si no tienes el archivo Excel"""
    # Datos C.3 (Tasa de paro según ámbito geográfico y sexo)
    data_c3 = {
        "Año": [2026, 2025, 2025, 2025, 2025, 2024, 2024, 2024, 2024],
        "Trimestre": ["I Trimestre", "IV Trimestre", "III Trimestre", "II Trimestre", "I Trimestre",
                      "IV Trimestre", "III Trimestre", "II Trimestre", "I Trimestre"],
        "Ciudad_Madrid_Ambos": [7.70, 6.61, 7.86, 6.91, 8.98, 7.64, 9.71, 8.29, 8.73],
        "Ciudad_Madrid_Hombres": [7.17, 6.61, 8.06, 6.36, 8.76, 8.00, 10.33, 7.97, 9.83],
        "Ciudad_Madrid_Mujeres": [8.23, 6.60, 7.67, 7.46, 9.20, 7.28, 9.08, 8.61, 7.61],
        "Comunidad_Madrid_Ambos": [7.91, 7.04, 7.98, 7.74, 9.11, 8.58, 9.65, 8.48, 9.19],
        "Comunidad_Madrid_Hombres": [6.89, 6.84, 7.63, 7.30, 9.47, 8.39, 9.01, 7.36, 9.18],
        "Comunidad_Madrid_Mujeres": [8.96, 7.25, 8.33, 8.19, 8.74, 8.77, 10.31, 9.64, 9.20],
        "España_Ambos": [10.83, 9.93, 10.45, 10.29, 11.36, 10.61, 11.21, 11.27, 12.29],
        "España_Hombres": [9.47, 8.76, 8.97, 9.20, 10.14, 9.53, 10.02, 10.08, 11.00],
        "España_Mujeres": [12.35, 11.24, 12.11, 11.50, 12.72, 11.83, 12.54, 12.61, 13.73]
    }

    data_31 = {
        "Año": [2026, 2025, 2025, 2025, 2025, 2024, 2024, 2024, 2024],
        "Trimestre": ["I Trimestre", "IV Trimestre", "III Trimestre", "II Trimestre", "I Trimestre",
                      "IV Trimestre", "III Trimestre", "II Trimestre", "I Trimestre"],
        "Ambos_Total": [7.70, 6.61, 7.86, 6.91, 8.98, 7.64, 9.71, 8.29, 8.73],
        "Ambos_Menos25": [17.54, 17.35, 25.20, 18.03, 19.43, 13.53, 27.88, 16.53, 15.19],
        "Ambos_25a54": [6.95, 5.29, 6.02, 5.03, 7.48, 6.42, 7.97, 7.54, 8.16],
        "Ambos_55ymas": [7.25, 7.97, 8.85, 9.58, 10.57, 10.11, 10.16, 8.40, 8.70],
        "Hombres_Total": [7.17, 6.61, 8.06, 6.36, 8.76, 8.00, 10.33, 7.97, 9.83],
        "Hombres_Menos25": [20.75, 15.09, 31.40, 15.00, 19.11, 14.90, 36.41, 16.99, 18.08],
        "Hombres_25a54": [6.30, 5.87, 5.89, 4.82, 7.56, 7.05, 8.29, 7.00, 9.42],
        "Hombres_55ymas": [5.79, 6.80, 8.50, 8.56, 9.39, 9.20, 9.34, 8.20, 8.67],
        "Mujeres_Total": [8.23, 6.60, 7.67, 7.46, 9.20, 7.28, 9.08, 8.61, 7.61],
        "Mujeres_Menos25": [14.07, 19.34, 19.45, 21.29, 19.74, 12.16, 19.14, 15.92, 12.21],
        "Mujeres_25a54": [7.60, 4.70, 6.15, 5.24, 7.39, 5.79, 7.64, 8.07, 6.88],
        "Mujeres_55ymas": [8.64, 9.11, 9.20, 10.57, 11.77, 11.02, 11.01, 8.60, 8.73]
    }

    return pd.DataFrame(data_c3), pd.DataFrame(data_31)

# Crear datos de ejemplo
df_c3, df_31 = crear_datos_ejemplo()

print("Datos cargados correctamente")
print(f"df_c3: {df_c3.shape}")
print(f"df_31: {df_31.shape}")

# ============================================================
# 4. Definir controles (igual que antes)
# ============================================================
años = sorted(df_c3["Año"].unique())
trimestres = df_c3["Trimestre"].unique().tolist()

# Selectores para C.3
selector_c3_ambito = widgets.Dropdown(
    options=["Ciudad de Madrid", "Comunidad de Madrid", "España"],
    value="Ciudad de Madrid",
    description="Ámbito:",
    style={'description_width': 'initial'}
)
selector_c3_sexo = widgets.Dropdown(
    options=["Ambos sexos", "Hombres", "Mujeres"],
    value="Ambos sexos",
    description="Sexo:",
    style={'description_width': 'initial'}
)

# Selectores para 3.1
selector_31_sexo = widgets.Dropdown(
    options=["Ambos sexos", "Hombres", "Mujeres"],
    value="Ambos sexos",
    description="Sexo:",
    style={'description_width': 'initial'}
)
selector_31_edad = widgets.Dropdown(
    options=["Total", "Menos de 25 años", "De 25 a 54 años", "De 55 años y más"],
    value="Total",
    description="Grupo de edad:",
    style={'description_width': 'initial'}
)

# Selectores para gráficos de secuencia C.3
selector_seq_c3_ambito = widgets.SelectMultiple(
    options=["Ciudad de Madrid", "Comunidad de Madrid", "España"],
    value=["Ciudad de Madrid", "Comunidad de Madrid", "España"],
    description="Ámbitos:",
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px', height='100px')
)
selector_seq_c3_sexo = widgets.SelectMultiple(
    options=["Ambos sexos", "Hombres", "Mujeres"],
    value=["Ambos sexos", "Hombres", "Mujeres"],
    description="Sexos:",
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px', height='80px')
)

# Selectores para gráficos de secuencia 3.1
selector_seq_31_sexo = widgets.SelectMultiple(
    options=["Ambos sexos", "Hombres", "Mujeres"],
    value=["Ambos sexos", "Hombres", "Mujeres"],
    description="Sexos:",
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px', height='80px')
)
selector_seq_31_edad = widgets.SelectMultiple(
    options=["Total", "Menos de 25 años", "De 25 a 54 años", "De 55 años y más"],
    value=["Total", "Menos de 25 años", "De 25 a 54 años", "De 55 años y más"],
    description="Grupos de edad:",
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px', height='100px')
)

# Selectores de año y trimestre
selector_año = widgets.Dropdown(
    options=años,
    value=años[0],
    description="Año:",
    style={'description_width': 'initial'}
)
selector_trimestre = widgets.Dropdown(
    options=trimestres,
    value=trimestres[0],
    description="Trimestre:",
    style={'description_width': 'initial'}
)

# ============================================================
# 5. Funciones de actualización (igual que antes)
# ============================================================
def actualizar_tabla_c3(año, trimestre, ambito, sexo):
    filtro = (df_c3["Año"] == año) & (df_c3["Trimestre"] == trimestre)
    if not any(filtro):
        return "No hay datos para los filtros seleccionados"

    col_map = {
        "Ciudad de Madrid": {"Ambos sexos": "Ciudad_Madrid_Ambos", "Hombres": "Ciudad_Madrid_Hombres", "Mujeres": "Ciudad_Madrid_Mujeres"},
        "Comunidad de Madrid": {"Ambos sexos": "Comunidad_Madrid_Ambos", "Hombres": "Comunidad_Madrid_Hombres", "Mujeres": "Comunidad_Madrid_Mujeres"},
        "España": {"Ambos sexos": "España_Ambos", "Hombres": "España_Hombres", "Mujeres": "España_Mujeres"}
    }
    col = col_map[ambito][sexo]
    valor = df_c3.loc[filtro, col].values[0]
    return f"Tasa de paro: {valor:.2f}%"

def actualizar_tabla_31(año, trimestre, sexo, edad):
    filtro = (df_31["Año"] == año) & (df_31["Trimestre"] == trimestre)
    if not any(filtro):
        return "No hay datos para los filtros seleccionados"

    col_map = {
        "Ambos sexos": {"Total": "Ambos_Total", "Menos de 25 años": "Ambos_Menos25", "De 25 a 54 años": "Ambos_25a54", "De 55 años y más": "Ambos_55ymas"},
        "Hombres": {"Total": "Hombres_Total", "Menos de 25 años": "Hombres_Menos25", "De 25 a 54 años": "Hombres_25a54", "De 55 años y más": "Hombres_55ymas"},
        "Mujeres": {"Total": "Mujeres_Total", "Menos de 25 años": "Mujeres_Menos25", "De 25 a 54 años": "Mujeres_25a54", "De 55 años y más": "Mujeres_55ymas"}
    }
    col = col_map[sexo][edad]
    valor = df_31.loc[filtro, col].values[0]
    return f"Tasa de paro: {valor:.2f}%"

def grafico_secuencia_c3(ambitos, sexos):
    fig = go.Figure()
    colores = ["#1f77b4", "#ff7f0e", "#2ca02c"]
    for i, ambito in enumerate(ambitos):
        for j, sexo in enumerate(sexos):
            col_map = {
                "Ciudad de Madrid": {"Ambos sexos": "Ciudad_Madrid_Ambos", "Hombres": "Ciudad_Madrid_Hombres", "Mujeres": "Ciudad_Madrid_Mujeres"},
                "Comunidad de Madrid": {"Ambos sexos": "Comunidad_Madrid_Ambos", "Hombres": "Comunidad_Madrid_Hombres", "Mujeres": "Comunidad_Madrid_Mujeres"},
                "España": {"Ambos sexos": "España_Ambos", "Hombres": "España_Hombres", "Mujeres": "España_Mujeres"}
            }
            col = col_map[ambito][sexo]
            df_c3["Tiempo"] = df_c3["Año"].astype(str) + " " + df_c3["Trimestre"]
            trimestre_order = {"I Trimestre": 1, "II Trimestre": 2, "III Trimestre": 3, "IV Trimestre": 4}
            df_plot = df_c3[["Tiempo", "Año", "Trimestre", col]].copy()
            df_plot["Trimestre_Num"] = df_plot["Trimestre"].map(trimestre_order)
            df_plot = df_plot.sort_values(["Año", "Trimestre_Num"])

            nombre = f"{ambito} - {sexo}"
            color = colores[i % len(colores)] if len(sexos) == 1 else colores[j % len(colores)]
            fig.add_trace(go.Scatter(
                x=df_plot["Tiempo"],
                y=df_plot[col],
                mode="lines+markers",
                name=nombre,
                line=dict(color=color)
            ))

    fig.update_layout(
        title="Evolución de la tasa de paro según ámbito geográfico y sexo",
        xaxis_title="Trimestre",
        yaxis_title="Tasa de paro (%)",
        height=400,
        legend=dict(x=1.02, y=1, xanchor="left", yanchor="top"),
        margin=dict(l=50, r=200, t=50, b=50)
    )
    return fig

def grafico_secuencia_31(sexos, edades):
    fig = go.Figure()
    colores = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]
    for i, sexo in enumerate(sexos):
        for j, edad in enumerate(edades):
            col_map = {
                "Ambos sexos": {"Total": "Ambos_Total", "Menos de 25 años": "Ambos_Menos25", "De 25 a 54 años": "Ambos_25a54", "De 55 años y más": "Ambos_55ymas"},
                "Hombres": {"Total": "Hombres_Total", "Menos de 25 años": "Hombres_Menos25", "De 25 a 54 años": "Hombres_25a54", "De 55 años y más": "Hombres_55ymas"},
                "Mujeres": {"Total": "Mujeres_Total", "Menos de 25 años": "Mujeres_Menos25", "De 25 a 54 años": "Mujeres_25a54", "De 55 años y más": "Mujeres_55ymas"}
            }
            col = col_map[sexo][edad]
            df_31["Tiempo"] = df_31["Año"].astype(str) + " " + df_31["Trimestre"]
            trimestre_order = {"I Trimestre": 1, "II Trimestre": 2, "III Trimestre": 3, "IV Trimestre": 4}
            df_plot = df_31[["Tiempo", "Año", "Trimestre", col]].copy()
            df_plot["Trimestre_Num"] = df_plot["Trimestre"].map(trimestre_order)
            df_plot = df_plot.sort_values(["Año", "Trimestre_Num"])

            nombre = f"{sexo} - {edad}"
            color = colores[j % len(colores)]
            fig.add_trace(go.Scatter(
                x=df_plot["Tiempo"],
                y=df_plot[col],
                mode="lines+markers",
                name=nombre,
                line=dict(color=color)
            ))

    fig.update_layout(
        title="Evolución de la tasa de paro según sexo y grupo de edad",
        xaxis_title="Trimestre",
        yaxis_title="Tasa de paro (%)",
        height=400,
        legend=dict(x=1.02, y=1, xanchor="left", yanchor="top"),
        margin=dict(l=50, r=200, t=50, b=50)
    )
    return fig

# ============================================================
# 6. Widgets de salida
# ============================================================
output_tabla_c3 = widgets.Output()
output_tabla_31 = widgets.Output()
output_grafico_c3 = widgets.Output()
output_grafico_31 = widgets.Output()

# ============================================================
# 7. Actualizar todos los outputs
# ============================================================
def actualizar_todo(change=None):
    with output_tabla_c3:
        clear_output()
        año = selector_año.value
        trimestre = selector_trimestre.value
        ambito = selector_c3_ambito.value
        sexo = selector_c3_sexo.value
        print(actualizar_tabla_c3(año, trimestre, ambito, sexo))

    with output_tabla_31:
        clear_output()
        año = selector_año.value
        trimestre = selector_trimestre.value
        sexo = selector_31_sexo.value
        edad = selector_31_edad.value
        print(actualizar_tabla_31(año, trimestre, sexo, edad))

    with output_grafico_c3:
        clear_output()
        fig = grafico_secuencia_c3(list(selector_seq_c3_ambito.value), list(selector_seq_c3_sexo.value))
        fig.show()

    with output_grafico_31:
        clear_output()
        fig = grafico_secuencia_31(list(selector_seq_31_sexo.value), list(selector_seq_31_edad.value))
        fig.show()

# Conectar eventos
selector_año.observe(actualizar_todo, names="value")
selector_trimestre.observe(actualizar_todo, names="value")
selector_c3_ambito.observe(actualizar_todo, names="value")
selector_c3_sexo.observe(actualizar_todo, names="value")
selector_31_sexo.observe(actualizar_todo, names="value")
selector_31_edad.observe(actualizar_todo, names="value")
selector_seq_c3_ambito.observe(actualizar_todo, names="value")
selector_seq_c3_sexo.observe(actualizar_todo, names="value")
selector_seq_31_sexo.observe(actualizar_todo, names="value")
selector_seq_31_edad.observe(actualizar_todo, names="value")

# ============================================================
# 8. Mostrar dashboard
# ============================================================
# Título
display(widgets.HTML("<h1>Dashboard de Tasas de Paro - EPA</h1>"))

# Filtros globales
display(widgets.HTML("<h3>Filtros globales</h3>"))
display(widgets.HBox([selector_año, selector_trimestre]))

# Tabla C.3
display(widgets.HTML("<h3>1. Tasa de paro según ámbito geográfico y sexo</h3>"))
display(widgets.HBox([selector_c3_ambito, selector_c3_sexo]))
display(output_tabla_c3)

# Tabla 3.1
display(widgets.HTML("<h3>2. Tasa de paro según sexo y grupo de edad</h3>"))
display(widgets.HBox([selector_31_sexo, selector_31_edad]))
display(output_tabla_31)

# Gráfico de secuencia C.3
display(widgets.HTML("<h3>3. Evolución de la tasa de paro según ámbito geográfico y sexo</h3>"))
display(widgets.HBox([selector_seq_c3_ambito, selector_seq_c3_sexo]))
display(output_grafico_c3)

# Gráfico de secuencia 3.1
display(widgets.HTML("<h3>4. Evolución de la tasa de paro según sexo y grupo de edad</h3>"))
display(widgets.HTML("<b>Seleccione sexos y grupos de edad:</b>"))
display(widgets.HBox([selector_seq_31_sexo, selector_seq_31_edad]))
display(output_grafico_31)

# Inicializar
actualizar_todo()
```
